```python
#16
from collections import Counter

def is_beautiful(a):
    cnt = Counter(a)
    for x in cnt:
        if cnt[x] != x:
            return False
    return True
```

```python
#17
def majority_element(a):
    cand, cnt = None, 0
    for x in a:
        if cnt == 0:
            cand = x
            cnt = 1
        elif x == cand:
            cnt += 1
        else:
            cnt -= 1
    return cand if a.count(cand) > len(a)//2 else None
```

```python
#18
def sort_even_odd(a):
    even = sorted([x for x in a if x % 2 == 0])
    odd = sorted([x for x in a if x % 2 == 1], reverse=True)
    res, i, j = [], 0, 0
    for x in a:
        if x % 2 == 0:
            res.append(even[i]); i += 1
        else:
            res.append(odd[j]); j += 1
    return res

```

```python
#19
def merge(a, b):
    i = j = 0
    res = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i]); i += 1
        else:
            res.append(b[j]); j += 1
    return res + a[i:] + b[j:]
```

```python
#20
def sum_k_digits(nums, k):
    return sum(x for x in nums if len(str(x)) == k)
```

```python
#21
def day_of_year(d, m):
    days = [31,29,31,30,31,30,31,31,30,31,30,31]
    if m < 1 or m > 12 or d < 1 or d > days[m-1]:
        return "invalid"
    return sum(days[:m-1]) + d
```

```python
#22
from collections import Counter

def most_common_name(names):
    cnt = Counter(names)
    mx = max(cnt.values())
    return min(name for name in cnt if cnt[name] == mx)
```

```python
#23
def count_words(s):
    return sum(1 for w in s.split() if len(w) >= 2)
```

```python
#24
def remove_duplicates(s):
    res = s[0]
    for c in s[1:]:
        if c != res[-1]:
            res += c
    return res
```

```python
#25
def max_k_digits(nums, k):
    vals = [x for x in nums if len(str(abs(x))) == k]
    return max(vals) if vals else None
```

```python
#26
import re

def sum_avg(s):
    nums = list(map(int, re.findall(r'\d+', s)))
    return sum(nums), sum(nums)/len(nums)
```

```python
#27
def valid_parentheses(s):
    st = []
    mp = {')':'(', ']':'[', '}':'{'}
    for c in s:
        if c in '([{':
            st.append(c)
        elif c in mp:
            if not st or st.pop() != mp[c]:
                return False
    return not st
```

```python
#28
def primes(n):
    p = [True]*(n+1)
    p[0]=p[1]=False
    for i in range(2,int(n**0.5)+1):
        if p[i]:
            for j in range(i*i,n+1,i):
                p[j]=False
    return [i for i in range(2,n+1) if p[i]]

def longest_consecutive_primes(n):
    ps = primes(n)
    best = []
    cur = [ps[0]]
    for i in range(1,len(ps)):
        if ps[i] == ps[i-1]+1:
            cur.append(ps[i])
        else:
            if len(cur) > len(best): best = cur
            cur = [ps[i]]
    return best if len(best)>=len(cur) else cur
```

```python
#29
def longest_increasing(a):
    best = cur = [a[0]]
    for i in range(1,len(a)):
        if a[i] == a[i-1] + 1:
            cur.append(a[i])
        else:
            if len(cur) > len(best): best = cur
            cur = [a[i]]
    return best if len(best)>=len(cur) else cur
```

```python
#30
def longest_non_consecutive(a):
    best = cur = [a[0]]
    for i in range(1,len(a)):
        if a[i] != a[i-1] + 1:
            cur.append(a[i])
        else:
            if len(cur) > len(best): best = cur
            cur = [a[i]]
    return best if len(best)>=len(cur) else cur
```

```python
#31
def min_by_swap(s):
    s = list(s)
    n = len(s)
    res = s[:]
    for i in range(n):
        for j in range(i+1,n):
            t = s[:]
            t[i], t[j] = t[j], t[i]
            if t[0] != '0' and t < res:
                res = t
    return ''.join(res)
```
