```python
#1
def reverse(path: str) -> int:
    try:
        with open(path, 'r') as f:
            content = f.read().strip()
            n = int(content)
            if n <= 0:
                raise ValueError("So khong phai so nguyen duong")
    except FileNotFoundError:
        print("Khong tim thay tep")
        return -1
    except ValueError:
        print("Noi dung tep khong hop le")
        return -1

    rev = int(str(n)[::-1])
    return rev
path = input().strip()
print(reverse(path))
```

```python
#2
def canWinNim(path: str) -> bool:
    try:
        with open(path, 'r') as f:
            n = int(f.read().strip())
            if n <= 0:
                raise ValueError
    except FileNotFoundError:
        print("Khong tim thay tep")
        return False
    except ValueError:
        print("Noi dung tep khong hop le")
        return False

    return n % 4 != 0
path = input().strip()
print(canWinNim(path))
```

```python
#3
def housesOfHogwarts(path: str):
    try:
        with open(path, 'r', encoding='utf-8') as f:
            lines = f.read().strip().splitlines()

            n = int(lines[0])      # số lượng tên
            names = lines[1:]      # danh sách tên

            if len(names) != n:
                print("So luong ten khong khop")
                return

            for name in names:
                print(name)

    except FileNotFoundError:
        print("Khong tim thay tep")
    except ValueError:
        print("Du lieu trong tep khong hop le")

```

```python
#4
def areAnagrams(path1: str, path2: str) -> bool:
    try:
        with open(path1, 'r', encoding='utf-8') as f1:
            s1 = f1.read()
        with open(path2, 'r', encoding='utf-8') as f2:
            s2 = f2.read()
    except FileNotFoundError:
        print("Khong tim thay tep")
        return False

    
    def normalize(s):
        s = s.lower()
        letters = []
        for ch in s:
            if ch.isalpha():  
                letters.append(ch)
        return sorted(letters)

    return normalize(s1) == normalize(s2)
path1 = input().strip()
path2 = input().strip()

print (areAnagrams(path1, path2))
```

```python
#5
def count(path: str, character: str) -> int:
    try:
        with open(path, 'r', encoding='utf-8') as f:
            text = f.read()
    except FileNotFoundError:
        print("Khong tim thay tep")
        return -1

    text = text.lower()
    character = character.lower()

    return text.count(character)


path = input().strip()
ch = input().strip()

print(count(path, ch))
```

```python
#6
def countCharacter(path: str, letter: str) -> int:
    try:
        with open(path, 'r', encoding='utf-8') as f:
            text = f.read()
    except FileNotFoundError:
        print("Khong tim thay tep")
        return -1

    text = text.lower()
    letter = letter.lower()

    count = 0
    for ch in text:
        if ch == letter:
            count += 1

    return count


path = input().strip()
letter = input().strip()

print(countCharacter(path, letter))
```

```python
#8
def fibonacciString(n):
    f = ["a", "b"]
    for i in range(2, n + 1):
        f.append(f[i-1] + f[i-2])
    return f

n = int(input())
fib = fibonacciString(n)

with open("output.txt", "w", encoding="utf-8") as f:
    for i in range(n + 1):
        f.write(f"f({i}) = {fib[i]}\n")
```

```python
#10
def rob(houses: list[int]) -> int:
    n = len(houses)
    if n == 0:
        return 0
    if n == 1:
        return houses[0]

    dp = [0] * n
    dp[0] = houses[0]
    dp[1] = max(houses[0], houses[1])

    for i in range(2, n):
        dp[i] = max(dp[i-1], dp[i-2] + houses[i])

    return dp[-1]
```

```python
#12
def maximumProduct(path: str) -> int:
    with open(path, "r") as f:
        n = int(f.readline())
        arr = list(map(int, f.readline().split()))

    arr.sort()
    return max(arr[-1]*arr[-2]*arr[-3],
               arr[0]*arr[1]*arr[-1])


path = input().strip()
print(maximumProduct(path))
```

```python
#13
def moveZeroes(path: str):
    with open(path, "r") as f:
        n = int(f.readline())
        arr = list(map(int, f.readline().split()))

    res = [x for x in arr if x != 0]
    res.extend([0] * (n - len(res)))
    return res


path = input().strip()
print(moveZeroes(path))
```

```python
#14
def averageTime(path: str):
    with open(path, "r") as f:
        lines = f.read().splitlines()[1:]

    times = []
    for line in lines:
        times.append(float(line.split(",")[1]))

    avg = sum(times) / len(times)
    return round(avg, 2)


path = input().strip()
print(averageTime(path))
```

```python
#15
def findMovies(path: str) -> list[str] | str:
    result = []

    with open(path, "r", encoding="utf-8") as f:
        lines = f.readlines()

    # Bỏ dòng tiêu đề
    for line in lines[1:]:
        parts = [p.strip() for p in line.split(",")]

        movie = parts[1]
        description = parts[2].lower()
        rating = float(parts[3])

        if rating > 8.0 and description != "boring":
            result.append(movie)

    if not result:
        return "All are boring!"

    return result
```

```python
#16
import math

def printPerimeters(path: str):
    with open(path, "r") as f:
        lines = [line.strip() for line in f if line.strip() != ""]

    i = 0
    while i < len(lines):
        shape_type = lines[i]
        data = lines[i + 1]
        i += 2

        if shape_type == "TYPE: SQUARE":
            side = float(data.split(":")[1])
            perimeter = 4 * side

        elif shape_type == "TYPE: RECTANGLE":
            parts = data.split()
            width = float(parts[1])
            height = float(parts[3])
            perimeter = 2 * (width + height)

        elif shape_type == "TYPE: CIRCLE":
            radius = float(data.split(":")[1])
            perimeter = 2 * math.pi * radius

        else:
            continue

        print(f"{perimeter:.2f}")

```

```python
#17
def productExceptSelf(path: str):
    with open(path, "r") as f:
        n = int(f.readline())
        nums = list(map(int, f.readline().split()))

    res = []
    for i in range(n):
        prod = 1
        for j in range(n):
            if i != j:
                prod *= nums[j]
        res.append(prod)
    return res


path = input().strip()
print(productExceptSelf(path))
```

```python
#18
import math

def estimatedTime(path: str):
    with open(path, "r") as f:
        lines = f.read().splitlines()

    sx, sy = map(float, lines[0].split(":")[1].split())
    dx, dy = map(float, lines[1].split(":")[1].split())
    v = float(lines[2].split(":")[1].split()[0])

    distance = math.sqrt((dx - sx)**2 + (dy - sy)**2)
    time = distance / v
    return round(time, 2)


path = input().strip()
print(estimatedTime(path))
```

```python
#20
def imageSmoother(path: str):
    with open(path, "r") as f:
        mat = [list(map(int, line.split())) for line in f]

    m, n = len(mat), len(mat[0])
    res = [[0]*n for _ in range(m)]

    for i in range(m):
        for j in range(n):
            s = 0
            cnt = 0
            for x in range(i-1, i+2):
                for y in range(j-1, j+2):
                    if 0 <= x < m and 0 <= y < n:
                        s += mat[x][y]
                        cnt += 1
            res[i][j] = s // cnt

    for row in res:
        print(*row)
```
