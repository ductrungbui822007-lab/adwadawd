```python
#1
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

arr = list(map(int, input().split()))
x = int(input())

print(binary_search(arr, 0, len(arr) - 1, x))
```

```python
#2
def count_occurrences(arr, x):
    return arr.count(x)

arr = list(map(int, input().split()))
x = int(input())

print(count_occurrences(arr, x))

```

```python
#3
arr = list(map(int, input().split()))
n = len(arr)

for i in range(n):
    for j in range(n - i - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]

print(*arr)
```

```python
#4
arr = list(map(int, input().split()))

freq = {}
for x in arr:
    freq[x] = freq.get(x, 0) + 1

max_count = max(freq.values())

for x in arr:
    if freq[x] == max_count:
        print(x, max_count)
        break
```

```python
#5
arr = list(map(int, input().split()))
X = int(input())

count = 0
seen = {}

for num in arr:
    if X - num in seen:
        count += seen[X - num]
    seen[num] = seen.get(num, 0) + 1

print(count)
```

```python
#6
def flatten(lst):
    res = []
    for item in lst:
        if isinstance(item, list):
            res.extend(flatten(item))
        else:
            res.append(item)
    return res

lst = eval(input())   # nhập dạng [[1,2],[3,[4,5]]]
print(flatten(lst))
```

```python
#7
nums = list(map(int, input().split()))
n = len(nums)

dp = [1] * n
for i in range(n):
    for j in range(i):
        if nums[j] < nums[i]:
            dp[i] = max(dp[i], dp[j] + 1)

print(max(dp))
```

```python
#8
intervals = eval(input())  # [[1,4],[2,6]]
queries = eval(input())    # [[2,3],[1,5]]

res = []
for ql, qr in queries:
    best = 10**18
    for l, r in intervals:
        if l <= ql and r >= qr:
            best = min(best, r - l + 1)
    res.append(best if best != 10**18 else -1)

print(res)
```

```python
#9
nums = list(map(int, input().split()))
n = len(nums)
res = []

for i in range(n):
    cnt = 0
    for j in range(i + 1, n):
        if nums[j] < nums[i]:
            cnt += 1
    res.append(cnt)

print(res)
```

```python
#10
nums = list(map(int, input().split()))
m = int(input())

max_sum = 0
prefix = 0
prefixes = []

for num in nums:
    prefix = (prefix + num) % m
    max_sum = max(max_sum, prefix)
    for p in prefixes:
        max_sum = max(max_sum, (prefix - p + m) % m)
    prefixes.append(prefix)

print(max_sum)
```
