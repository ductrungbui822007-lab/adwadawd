```python
#1
def merge_lists(a, b):
    i = j = 0
    res = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i]); i += 1
        else:
            res.append(b[j]); j += 1
    return res + a[i:] + b[j:]
```

```python
#2
def kth_largest(matrix, k):
    arr = []
    for row in matrix:
        arr.extend(row)
    arr.sort(reverse=True)
    return arr[k-1]
```

```python
#3
def min_by_swap(s):
    s = list(s)
    best = s[:]
    n = len(s)
    for i in range(n):
        for j in range(i+1, n):
            t = s[:]
            t[i], t[j] = t[j], t[i]
            if t[0] != '0' and t < best:
                best = t
    return ''.join(best)
```

```python
#4
def valid_abc(s):
    if not s: return "NO"
    if any(c not in "ABC" for c in s): return "NO"
    if s != ''.join(sorted(s)): return "NO"
    return "YES"
```

```python
#5
def is_subarray(a, b):
    for i in range(len(a) - len(b) + 1):
        if a[i:i+len(b)] == b:
            return "YES"
    return "NO"
```

```python
#6
def next_permutation(a):
    n = len(a)
    i = n - 2
    while i >= 0 and a[i] >= a[i+1]:
        i -= 1
    if i == -1:
        return a[::-1]

    j = n - 1
    while a[j] <= a[i]:
        j -= 1
    a[i], a[j] = a[j], a[i]
    a[i+1:] = reversed(a[i+1:])
    return a
```

```python
#7
def canPlaceFlowers(flowerbed, n):
    count = 0
    for i in range(len(flowerbed)):
        if flowerbed[i] == 0:
            left = (i == 0 or flowerbed[i-1] == 0)
            right = (i == len(flowerbed)-1 or flowerbed[i+1] == 0)
            if left and right:
                flowerbed[i] = 1
                count += 1
    return count >= n

```

```python
#8
def isRobotBounded(moves):
    x = y = 0
    for c in moves:
        if c == 'U': y += 1
        elif c == 'D': y -= 1
        elif c == 'L': x -= 1
        elif c == 'R': x += 1
    return x == 0 and y == 0
```

```python
#9
def median(nums):
    nums = nums[:]   # cho phép xử lý
    nums.sort()
    n = len(nums)
    if n % 2 == 1:
        return nums[n//2]
    return (nums[n//2-1] + nums[n//2]) // 2
```

```python
#10
def findRadius(houses, heaters):
    heaters.sort()
    res = 0
    for h in houses:
        l, r = 0, len(heaters)-1
        while l < r:
            m = (l + r) // 2
            if heaters[m] < h:
                l = m + 1
            else:
                r = m
        d1 = abs(heaters[l] - h)
        d2 = abs(heaters[l-1] - h) if l > 0 else 10**18
        res = max(res, min(d1, d2))
    return res
```
